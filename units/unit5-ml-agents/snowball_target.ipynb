Change Kernel: Runtime -> Change runtime type -> Python 3.10, Restart Runtime, When you register a new kernel inside Colab (ipykernel install --user), the UI does show "Python 3.10 (conda)". But Colab’s runtime sandbox does not actually persist the new kernel environment across restarts seeing "Restarting runtime..."
