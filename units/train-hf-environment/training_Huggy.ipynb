```python
%%writefile config/ppo/Huggy.yaml
behaviors:
  Huggy:
    trainer_type: ppo
    hyperparameters:
      batch_size: 2048
      buffer_size: 20480
      learning_rate: 0.0003
      beta: 0.005
      epsilon: 0.2
      lambd: 0.95
      num_epoch: 3
      learning_rate_schedule: linear
    network_settings:
      normalize: true
      hidden_units: 512
      num_layers: 3
      vis_encode_type: simple
    reward_signals:
      extrinsic:
        gamma: 0.995
        strength: 1.0
    checkpoint_interval: 200000
    keep_checkpoints: 15
    max_steps: 2e6
    time_horizon: 1000
    summary_freq: 50000
```

## Train the agent

Launch mlagents-learn and select executable containing the environment. ML Angents runs training script when setting four parameters:

* mlagents-learn <config>: the path where the hyperparameter config file is.
* --env: where the environment executable is.
* --run-id: the name you want to give to your training run id.
* --no-graphics: to not launch the visualization during the training.

NB: Training takes 30-45 mins, use --resume floag to continue training in case of interuption

```python
!mlagents-learn "./config/ppo/Huggy.yaml"
--env="./trained-envs-executables/linux/Huggy/Huggy.x86_64"
--run-id="Huggy"
--no-graphics
```
